```python
import json
import requests

TEAM_ROSTERS_DIR = "data/teams"
GAMES_FILE = "data/games.json"
ALL_SKATERS_FILE = "data/all_skaters.json"
ALL_GOALIES_FILE = "data/all_goalies.json"

GAMETYPE = "02" # Use "19" for Four Nations 
SEASON = "20242025"

# Individual Games
BOXSCORE_API_URL = "https://api-web.nhle.com/v1/gamecenter/{}/boxscore"
GAME_STORY_API_URL = "https://api-web.nhle.com/v1/wsc/game-story/{}"
# Goals, Assists, Plus/Minus, Powerplay Points, Game-Winning Goals
NHL_API_URL_SKATER_SUMMARY = "https://api.nhle.com/stats/rest/en/skater/summary?limit=-1&cayenneExp=seasonId={}%20and%20gameTypeId={}"
# Shots, Hits
NHL_API_URL_SKATER_REALTIME = "https://api.nhle.com/stats/rest/en/skater/realtime?limit=-1&cayenneExp=seasonId={}%20and%20gameTypeId={}"
# Saves, Goals Against
NHL_API_URL_GOALIE_SUMMARY = "https://api.nhle.com/stats/rest/en/goalie/summary?limit=-1&cayenneExp=seasonId={}%20and%20gameTypeId={}"

# Example Game
game_id = "2024030211"
```

```python
# Game Story
def request_game_story(game_id):
    """Fetch game story from JSON file."""
    response = requests.get(GAME_STORY_API_URL.format(game_id))
    if response.status_code == 200:
        story = response.json()
    else:
        return None
    if story.get("gameState") in ["FUT", "PRE"]:
        return None
    return story

story = request_game_story(game_id)
```
