```python
import json
import requests

TEAM_ROSTERS_DIR = "data/teams"
GAMES_FILE = "data/games.json"
ALL_SKATERS_FILE = "data/all_skaters.json"
ALL_GOALIES_FILE = "data/all_goalies.json"

GAMETYPE = "02" # Use "19" for Four Nations 
SEASON = "20242025"

# Individual Games
BOXSCORE_API_URL = "https://api-web.nhle.com/v1/gamecenter/{}/boxscore"
GAME_STORY_API_URL = "https://api-web.nhle.com/v1/wsc/game-story/{}"
GAME_PLAY_BY_PLAY_API_URL = "https://api-web.nhle.com/v1/gamecenter/{}/play-by-play"
# Goals, Assists, Plus/Minus, Powerplay Points, Game-Winning Goals
NHL_API_URL_SKATER_SUMMARY = "https://api.nhle.com/stats/rest/en/skater/summary?limit=-1&cayenneExp=seasonId={}%20and%20gameTypeId={}"
# Shots, Hits
NHL_API_URL_SKATER_REALTIME = "https://api.nhle.com/stats/rest/en/skater/realtime?limit=-1&cayenneExp=seasonId={}%20and%20gameTypeId={}"
# Saves, Goals Against
NHL_API_URL_GOALIE_SUMMARY = "https://api.nhle.com/stats/rest/en/goalie/summary?limit=-1&cayenneExp=seasonId={}%20and%20gameTypeId={}"

# Team Information
NHL_API_ALL_TEAMS = "https://api.nhle.com/stats/rest/en/team"
NHL_API_ALL_FRANCHISES = "https://api.nhle.com/stats/rest/en/franchise"

# Player Information
NHL_API_PLAYER_LANDING ="https://api-web.nhle.com/v1/player/{}/landing"

# Example Game
game_id = "2024030211"
player_id = "8478402" # Connor McDavid
```

```python
# Game Story
def request_game_story(game_id):
    """Fetch game story from JSON file."""
    response = requests.get(GAME_STORY_API_URL.format(game_id))
    if response.status_code == 200:
        story = response.json()
    else:
        return None
    if story.get("gameState") in ["FUT", "PRE"]:
        return None
    return story

def request_game_play_by_play(game_id):
    """Fetch game play-by-play from JSON file."""
    response = requests.get(GAME_PLAY_BY_PLAY_API_URL.format(game_id))
    if response.status_code == 200:
        play_by_play = response.json()
    else:
        return None
    if play_by_play.get("gameState") in ["FUT", "PRE"]:
        return None
    return play_by_play

def request_all_teams():
    """Fetch all teams from JSON file."""
    response = requests.get(NHL_API_ALL_TEAMS)
    if response.status_code == 200:
        teams = response.json()
        return teams.get("data", [])
    else:
        return []

def request_all_franchises():
    """Fetch all franchises from JSON file."""
    response = requests.get(NHL_API_ALL_FRANCHISES)
    if response.status_code == 200:
        franchises = response.json()
        return franchises.get("data", [])
    else:
        return []

def request_player_landing(player_id):
    """Fetch player landing page from JSON file."""
    response = requests.get(NHL_API_PLAYER_LANDING.format(player_id))
    if response.status_code == 200:
        player_landing = response.json()
        return player_landing
    else:
        return None

story = request_game_story(game_id)
# play_by_play = request_game_play_by_play(game_id)
# all_teams = request_all_teams()
# all_franchises = request_all_franchises()
mcdavid = request_player_landing(player_id)
```
